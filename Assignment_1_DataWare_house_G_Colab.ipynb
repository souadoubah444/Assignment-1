```python
gcs_uri = "gs://austin-bike-project-123456/Austin_MetroBike_Trips_20251204 (1).csv"
print("Reading from:", gcs_uri)

df = pd.read_csv(gcs_uri)

df.head()
```

```python

print(df.columns.tolist())

df = df.rename(columns={
    "Trip ID": "trip_id",
    "Membership or Pass Type": "membership_type",
    "Bicycle ID": "bicycle_id",
    "Bike Type": "bike_type",
    "Checkout Datetime": "checkout_datetime",
    "Checkout Date": "checkout_date",
    "Checkout Time": "checkout_time",
    "Checkout Kiosk ID": "checkout_kiosk_id",
    "Checkout Kiosk": "checkout_kiosk",
    "Return Kiosk ID": "return_kiosk_id",
    "Return Kiosk": "return_kiosk",
    "Trip Duration Minutes": "trip_duration_minutes",
    "Month": "month_raw",
    "Year": "year_raw"
})

df.head()
```

```python

df["checkout_datetime"] = pd.to_datetime(
    df["checkout_datetime"],
    errors="coerce",
    infer_datetime_format=True
)


df["trip_duration_minutes"] = pd.to_numeric(
    df["trip_duration_minutes"],
    errors="coerce"
)


df[["checkout_datetime", "trip_duration_minutes"]].head()
```

```python

df["checkout_datetime"] = pd.to_datetime(
    df["checkout_datetime"],
    errors="coerce",
    infer_datetime_format=True
)


df["trip_duration_minutes"] = pd.to_numeric(
    df["trip_duration_minutes"],
    errors="coerce"
)


df[["checkout_datetime", "trip_duration_minutes"]].head()
```

```python

df["return_datetime"] = df["checkout_datetime"] + pd.to_timedelta(
    df["trip_duration_minutes"], unit="m"
)


df["checkout_date"] = df["checkout_datetime"].dt.date
df["return_date"]   = df["return_datetime"].dt.date

df[["checkout_datetime", "checkout_date", "return_datetime", "return_date"]].head()
```

```python

df["year"]        = df["checkout_datetime"].dt.year
df["month"]       = df["checkout_datetime"].dt.month
df["day"]         = df["checkout_datetime"].dt.day
df["hour"]        = df["checkout_datetime"].dt.hour
df["day_of_week"] = df["checkout_datetime"].dt.day_name()
df["hour_of_day"] = df["checkout_datetime"].dt.hour

df["is_round_trip"] = (df["checkout_kiosk_id"] == df["return_kiosk_id"])

df[[
    "trip_id", "bicycle_id", "checkout_datetime", "return_datetime",
    "trip_duration_minutes", "day_of_week", "hour_of_day", "is_round_trip"
]].head()
```

```python

before = len(df)
df = df.drop_duplicates(subset=["trip_id"])
after = len(df)
print("Duplicates removed:", before - after)


for col in ["trip_id", "bicycle_id", "checkout_kiosk_id", "return_kiosk_id"]:
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")


critical_cols = [
    "trip_id", "checkout_datetime", "return_datetime",
    "checkout_kiosk_id", "return_kiosk_id", "bicycle_id"
]
before = len(df)
df = df.dropna(subset=critical_cols)
after = len(df)
print("Rows removed due to NULLs:", before - after)
print("Final row count:", after)
```

```python

df["date_key"] = df["checkout_datetime"].dt.strftime("%Y%m%d").astype(int)

dim_date = pd.DataFrame({
    "date_key": df["date_key"],
    "date": df["checkout_date"],
    "year": df["year"],
    "month": df["month"],
    "day": df["day"],
    "hour": df["hour"]
}).drop_duplicates().reset_index(drop=True)

dim_date.head()
```

```python

stations_checkout = df[["checkout_kiosk_id", "checkout_kiosk"]].rename(
    columns={"checkout_kiosk_id": "station_id", "checkout_kiosk": "station_name"}
)
stations_return = df[["return_kiosk_id", "return_kiosk"]].rename(
    columns={"return_kiosk_id": "station_id", "return_kiosk": "station_name"}
)

dim_station = pd.concat([stations_checkout, stations_return], ignore_index=True)
dim_station = dim_station.dropna(subset=["station_id"]).drop_duplicates().reset_index(drop=True)


dim_station["station_id"] = dim_station["station_id"].astype("Int64")

dim_station.head()
```

```python
dim_bike = df[["bicycle_id", "bike_type"]].drop_duplicates().reset_index(drop=True)
dim_bike = dim_bike.rename(columns={"bicycle_id": "bike_id"})
dim_bike["bike_id"] = dim_bike["bike_id"].astype("Int64")

dim_bike.head()
```

```python
dim_membership = df[["membership_type"]].drop_duplicates().reset_index(drop=True)
dim_membership["membership_key"] = range(1, len(dim_membership) + 1)

dim_membership.head()
```

```python
fact = df.copy()

fact = fact.merge(dim_bike, left_on="bicycle_id", right_on="bike_id", how="left")


fact = fact.merge(dim_membership, on="membership_type", how="left")


fact = fact.rename(columns={
    "checkout_kiosk_id": "station_start_id",
    "return_kiosk_id":   "station_end_id"
})


fact["trip_key"] = range(1, len(fact) + 1)


fact_trips = fact[[
    "trip_key",
    "trip_id",
    "date_key",
    "station_start_id",
    "station_end_id",
    "bike_id",
    "membership_key",
    "checkout_datetime",
    "return_datetime",
    "trip_duration_minutes",
    "day_of_week",
    "hour_of_day",
    "is_round_trip"
]]

fact_trips.head()
```
